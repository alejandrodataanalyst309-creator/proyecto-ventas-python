## Ejercicio 7 · Gráficas agrupadas con Seaborn \(13 min\)

Seaborn permite comparar grupos de forma más clara usando argumentos como `hue`.

Crearemos:

1. Ingreso total por categoría separado por canal.
2. Margen porcentual por canal separado por tipo de cliente.
3. Distribución de ingresos por canal y tipo de cliente.

### Espacio de trabajo del estudiante

**Mi interpretación de la consigna:**  

**Mi procedimiento / fórmula / consulta:**  

**Resultado obtenido:**  

**Conclusión breve:**  


````python
# Ingreso total por categoría separado por canal
resumen_categoria=df_limpio.groupby(['categoria','canal'],as_index=False).agg(ingreso_total=('ingreso','sum'))
plt.figure(figsize=(8,6))
sns.barplot(data=resumen_categoria,x='categoria',y='ingreso_total',hue='canal')
plt.title("Ingreso Total por categoria")
plt.xlabel('Categoría')
plt.ylabel('Ingreso Total')
plt.show()
````

````python
# Margen porcentual por canal separado por tipo de cliente.
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Agrupar por canal y tipo de cliente calculando la suma de ingresos y costos
resumen_margen = df_limpio.groupby(['canal', 'tipo_cliente'], as_index=False).agg(
    ingreso_total=('ingreso', 'sum'),
    costo_total=('costo', 'sum')
)

# 2. Calcular el margen porcentual (Margen / Ingresos)
resumen_margen['margen_pct'] = (resumen_margen['ingreso_total'] - resumen_margen['costo_total']) / resumen_margen['ingreso_total']

# 3. Crear el gráfico de barras con Seaborn usando 'hue' para separar por tipo de cliente
plt.figure(figsize=(9, 5))
sns.barplot(
    data=resumen_margen,
    x='canal',
    y='margen_pct',
    hue='tipo_cliente',
    palette='muted'
)

plt.title('Margen Porcentual por Canal y Tipo de Cliente')
plt.xlabel('Canal de Venta')
plt.ylabel('Margen Porcentual')
plt.ylim(0, 1)  # Ajusta el límite del eje Y entre 0% y 100% (0.0 a 1.0)
plt.legend(title='Tipo de Cliente')
plt.grid(True, axis='y')
plt.show()
````

````python
# Distribución de ingresos por canal y tipo de cliente.
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))

# Usamos boxplot para ver la distribución y dispersión de los ingresos
sns.boxplot(
    data=df_limpio,
    x='canal',
    y='ingreso',
    hue='tipo_cliente',
    palette='Set2'
)

plt.title('Distribución de Ingresos por Canal y Tipo de Cliente')
plt.xlabel('Canal de Venta')
plt.ylabel('Ingreso')
plt.legend(title='Tipo de Cliente')
plt.grid(True, axis='y')
plt.show()
````

## Ejercicio 8 · Outliers con IQR \(8 min\)

La regla IQR permite detectar valores inusualmente altos o bajos.

No significa que esos valores estén mal. Significa que deben revisarse antes de sacar conclusiones.

### Espacio de trabajo del estudiante

**Mi interpretación de la consigna:**  

**Mi procedimiento / fórmula / consulta:**  

**Resultado obtenido:**  

**Conclusión breve:**  


````python
# TODO estudiante: escribe aquí tu código.
# Función para calcular límites IQR
def calcular_limites_iqr(tabla,columna):
    q1=tabla[columna].quantile(0.25)
    q3=tabla[columna].quantile(0.75)
    iqr=q3-q1
    limite_inferior=q1-1.5*iqr
    limite_superior=q3+1.5*iqr
    return limite_inferior,limite_superior
````

````python
# Indentificar Ventas Atípicas por Ingreso
limite_inf,limite_sup=calcular_limites_iqr(df_limpio,'ingreso')
print('Límite inferior:', limite_inf)
print('Límite superior:', limite_sup)
````

````python

outliers_ingreso=df_limpio[(df_limpio['ingreso']<limite_inf) | (df_limpio['ingreso']>limite_sup)]
df_limpio[(df_limpio['ingreso']<limite_inf) | (df_limpio['ingreso']>limite_sup)]
````

````python
print('Cantidad de ventas atípicas:', len(outliers_ingreso))
#
````

## Ejercicio 9 · Segmentación de ventas \(8 min\)

Segmentar ayuda a convertir números en etiquetas interpretables. Vamos a clasificar cada venta por tamaño de ingreso.

### Espacio de trabajo del estudiante

**Mi interpretación de la consigna:**  

**Mi procedimiento / fórmula / consulta:**  

**Resultado obtenido:**  

**Conclusión breve:**  


````python
# Función para segmentación
def clasificar_ingreso(ingreso):
    if ingreso < 500000:
        return 'Bajo'
    elif ingreso <= 2000000:
        return 'Medio'
    else:
        return 'Alto'

# Aplicar la función al DataFrame limpio creando una nueva columna
df_limpio['segmento_ingreso'] = df_limpio['ingreso'].apply(clasificar_ingreso)

print("Segmentación aplicada correctamente.")
df_limpio[['venta_id', 'ingreso', 'segmento_ingreso']].head()
````

````python
# Conteo por Segmento
conteo_segmentos = df_limpio['segmento_ingreso'].value_counts().reset_index()
conteo_segmentos.columns = ['segmento', 'cantidad_ventas']

print(conteo_segmentos)
````
